```python
import re
import numpy as np

UNCERTAINTY = [
    "maybe", "perhaps", "probably", "i guess", "i think",
    "kind of", "sort of", "not sure", "i don't know",
    "i'm not sure", "i am not sure"
]

CONTRAST = [
    "but", "however", "although", "though", "yet",
    "on the other hand"
]

DESIRE = [
    "want", "wish", "hope", "would like", "intend",
    "plan", "should", "need to", "supposed to"
]

RESISTANCE = [
    "can't", "cannot", "won't", "avoid", "hard",
    "difficult", "struggle", "procrastinate", "never",
    "not able", "don't feel like"
]

FILLERS = [
    "um", "uh", "hmm", "er", "like", "you know"
]

def count_phrases(text, phrases):
    text = text.lower()
    return sum(len(re.findall(r"\b" + re.escape(p) + r"\b", text)) for p in phrases)

def count_repetitions(text):
    words = re.findall(r"\b\w+\b", text.lower())
    return sum(1 for i in range(1, len(words)) if words[i] == words[i-1])

def extract_hesitation_text_features(text):
    text = text.lower()

    n_uncertainty = count_phrases(text, UNCERTAINTY)
    n_contrast = count_phrases(text, CONTRAST)
    n_desire = count_phrases(text, DESIRE)
    n_resistance = count_phrases(text, RESISTANCE)
    n_fillers = count_phrases(text, FILLERS)
    n_repetitions = count_repetitions(text)

    n_words = max(1, len(re.findall(r"\b\w+\b", text)))

    return {
        "num_words": n_words,
        "num_uncertainty": n_uncertainty,
        "num_contrast": n_contrast,
        "num_desire": n_desire,
        "num_resistance": n_resistance,
        "num_fillers": n_fillers,
        "num_repetitions": n_repetitions,

        "rate_uncertainty": n_uncertainty / n_words,
        "rate_contrast": n_contrast / n_words,
        "rate_desire": n_desire / n_words,
        "rate_resistance": n_resistance / n_words,
        "rate_fillers": n_fillers / n_words,

        "desire_resistance_product": n_desire * n_resistance,
        "contrast_uncertainty_product": n_contrast * n_uncertainty
    }
```

```python
text = "I know I should exercise, but I don't know, it's hard and I guess I keep avoiding it."

features = extract_hesitation_text_features(text)
print(features)
```
